# Renfe Scraping Software

This notebook is intended to show the process of scraping Renfe data using the `RenfeScraper` class. The class is designed to scrape data from the Renfe website and save it in a structured format. Two output files are obtained:
- `trips.csv`: Contains information about the trips, including the service ID, trip ID, and other relevant details.
- `stops.csv`: Contains information about the stops, including the stop ID, arrival and departure times, and other relevant details.

## 0. Import Libraries

```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import datetime

from copy import deepcopy
from functools import cached_property
from typing import Dict, Mapping, List

from robin.scraping.constants import (
    PRICES_COLUMNS, RENFE_STATIONS_PATH, SPANISH_CORRIDOR_PATH, TIME_SLOT_SIZE
)
from robin.scraping.entities import DataLoader
from robin.supply.entities import RollingStock, Service, Supply
from robin.supply.utils import get_time
from robin.supply.saver.entities import SupplySaver
```

## 3. Load scraped data

The following cell loads the scraped data from the `trips.csv` and `stops.csv` files. The `DataLoader` class is used to load the data and create a supply entity.

```python
INFINITE_CAPACITY = 999999

def max_demand(capacity: int) -> int:
    """
    Returns the maximum demand for a given train based on the capacity.

    Args:
        capacity (int): The capacity of the rolling stock.
    Returns:
        int: The maximum demand for the train based on the capacity.
    """
    # return INFINITE_CAPACITY # Uncomment this line to set the maximum demand to infinite
    # return int(capacity * 1.5) # Uncomment this line to set the maximum demand to 1.5 times the capacity
    return capacity


class RSDataLoader(DataLoader):

    def __init__(
        self,
        stops_path: str,
        prices_path: str,
        seat_components: Mapping[int, int],
        renfe_stations_path: str = RENFE_STATIONS_PATH,
        spanish_corridor_path: str = SPANISH_CORRIDOR_PATH,
        time_slot_size: int = TIME_SLOT_SIZE
    ) -> None:
        """
        Initialize a DataLoader with the given paths to the CSV files.

        Args:
            stops_path (str): Path to the stops CSV file.
            prices_path (str): Path to the prices CSV file.
            seat_components (Mapping[int, int]): Dictionary with seat components.
            seat_quantity (Mapping[int, int]): Dictionary with seat quantity.
            renfe_stations_path (str, optional): Path to the Renfe stations CSV file.
            spanish_corridor_path (str, optional): Path to the Spanish corridor YAML file.
            time_slot_size (int, optional): Time slot size in minutes.
        """
        self.stops = pd.read_csv(stops_path, dtype={'stop_id': str})
        self.prices = pd.read_csv(prices_path, dtype={'origin': str, 'destination': str})
        self.renfe_stations = pd.read_csv(renfe_stations_path, delimiter=';', dtype={'ADIF_ID': str, 'RENFE_ID': str})
        self.seat_components = seat_components
        self.seat_names = self.prices.columns[PRICES_COLUMNS + 3:] # Skip the new columns added in the extended prices dataset
        self.spanish_corridor = self._read_yaml(path=spanish_corridor_path)
        self.time_slot_size = time_slot_size
    
    @cached_property
    def rolling_stocks(self) -> Dict[str, RollingStock]:
        """
        Returns a dictionary of rolling stocks with their IDs as keys.

        Returns:
            Dict[str, RollingStock]: Dictionary of rolling stocks with their IDs as keys.
        """
        df_rs = self.prices[['capacity', 'train_model']].drop_duplicates()
        rolling_stocks = {}
        for _, row in df_rs.iterrows():
            capacity=int(row['capacity'])
            _id = f'{row["train_model"]}_C{capacity}'
            rs = RollingStock(_id, _id, seats={1: max_demand(capacity)})
            rolling_stocks[_id] = rs
        return rolling_stocks
    
    @cached_property
    def services(self) -> List[Service]:
        """
        Returns a list of services.

        Returns:
            List[Service]: List of services.
        """
        trips = deepcopy(self.prices)
        trips['date'] = trips['service_id'].apply(lambda service_id: datetime.datetime.strptime(service_id.split('_')[1], '%d-%m-%Y-%H.%M').date())
        trips['line'] = trips['service_id'].apply(lambda service_id: self.lines[service_id.split('_')[0]])
        trips['train_service_provider'] = trips['tsp'].apply(
            lambda tsp_name: next(tsp for tsp in self.tsps.values() if tsp.name == tsp_name)
        )
        trips['time_slot'] = trips['service_id'].apply(
            lambda service_id: self.time_slots[
                f'{get_time(service_id.split("-")[-1].replace(".", ":") + ":00").seconds // 60}'
                f'{datetime.timedelta(minutes=self.time_slot_size).seconds // 60}'
            ]
        )
        trips['rolling_stock'] = trips['service_id'].apply(
            lambda service_id: self.rolling_stocks[
                f'{trips[trips["service_id"] == service_id]["train_model"].values[0]}_C{int(trips[trips["service_id"] == service_id]["capacity"].values[0])}'
            ]
        )
        trips['prices'] = trips['service_id'].apply(
            lambda service_id: self._get_trip_prices(
                service_id=service_id, line=trips['line'].values[0]
            )
        )
        trips['service'] = trips.apply(
            lambda service: Service(
                service['service_id'],
                service['date'],
                service['line'],
                service['train_service_provider'],
                service['time_slot'],
                service['rolling_stock'],
                service['prices']
            ),
            axis=1
        )
        return trips['service'].values.tolist()
```

```python
# The different seats with their hard and soft types
seat_components = {
    'BasicSeat': (1, 1)
}

data_loader = RSDataLoader(
    stops_path='../data/MAD-BCN/stop_times/stop_times_MAD-BCN_2025.csv',
    prices_path='../data/MAD-BCN/prices/prices_ext_MAD-BCN_2025.csv',
    seat_components=seat_components
)
```

## 4. Save supply entities

The following cell saves the supply entities to a YAML file. The `SupplySaver` class is used to save the data in a structured format. The `to_yaml` method is used to save the data to a file.

```python
supply_saver = SupplySaver(services=data_loader.services)
supply_saver.to_yaml(output_path='../data/MAD-BCN/supply_MAD-BCN_2025.yaml')
```
